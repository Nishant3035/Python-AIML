# Interview Quick Revision

| Topic | Remember |
|---|---|
| arange | Stop is excluded |
| linspace | Equal intervals |
| reshape | Elements must match |
| axis=0 | Columns |
| axis=1 | Rows |
| flatten | Returns a copy |
| ravel | Returns a view when possible |
| Boolean Indexing | Used heavily in preprocessing |
| Vectorization | Prefer over Python loops |
| Broadcasting | Operates on compatible shapes |
