```python
import numpy as np
arr = np.arange(1,26).reshape(5,5)
print(arr)
```

```python
# 1. Print the shape.
print(arr.shape)
```

```python
# 2. Print the third row.
print(arr[2,:])
```

```python
# 3. Print the second column.
print(arr[:,1])
```

```python
# 4. Print the bottom-right 2×2 matrix.
print(arr[3:,3:])
```

```python
# 5. Replace 13 with 1300.
arr[2,2] = 1300
print(arr)
```

```python
# 6. Convert the array to float64.
arr1 = arr.astype(np.float16)
print(arr1)
```

```python
# 7. Find the sum of every row.
row1 = np.sum(arr,axis = 1)
print(row1)
```

```python
# 8. Find the maximum value of every column.
max_col = np.max(arr,axis =0)
print(max_col)
```

```python
# 9. Print only even numbers using boolean indexing.
print(arr[arr%2 == 0])
```

```python
# 10. Flatten the array into a 1D array.
```
